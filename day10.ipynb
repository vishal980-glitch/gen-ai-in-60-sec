```
import math
def relu(x): return max(0, x)
def sigmoid(x):
    return 1 / (1 + math.exp(-x))
print(relu(-3), relu(5))
print(round(sigmoid(0), 2))
print(round(sigmoid(5), 2))
print(round(sigmoid(-5), 2))
```
