```
raw = ["  Hello ", "HELLO", "", "AI  ", "ai"]
clean = []
for t in raw:
    t = t.strip().lower()
    if t and t not in clean:
        clean.append(t)
print(clean)
print(len(raw), "->", len(clean))
```
