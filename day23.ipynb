# Day 23: Hallucination Kyun Hoti Hai?
Run the cell (Shift+Enter), then try the challenge.

**Challenge:** facts mein ek naya fact daalo. Comment your output on the video!

```
facts = {"india capital": "Delhi"}
def ai(q):
    if q in facts:
        return facts[q]
    return "Mumbai"
print(ai("india capital"))
print(ai("mars capital"))
print("AI ne confident jhooth bola!")
```
