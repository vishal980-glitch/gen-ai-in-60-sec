```
text = "main code likh raha hoon"
text += " main app bana raha hoon"
words = text.split()
nxt = {}
for a, b in zip(words, words[1:]):
    nxt.setdefault(a, []).append(b)
print("main →", nxt["main"])
print("raha →", nxt["raha"])
```
