```
w = "ai seekho ai banao ai seekho".split()
pairs = list(zip(w, w[1:]))
print(pairs[:3])
n = sum(1 for a, b in pairs if a == "ai")
for nxt in ("seekho", "banao"):
    c = pairs.count(("ai", nxt))
    print("ai ->", nxt, round(c / n, 2))
```
