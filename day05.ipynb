# Day 5: Tokens kya hote hain?

```
text = "ai seekho, ai banao"
vocab = {}
tokens = text.replace(",", "").split()
for t in tokens:
    if t not in vocab:
        vocab[t] = len(vocab)
ids = [vocab[t] for t in tokens]
print(vocab, ids)
```
